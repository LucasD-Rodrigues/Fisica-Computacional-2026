```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng()
```

```python
#1

def randomwalk(p, N):
    
    x = [0]

    for i in range(N):
        r = rng.random()
        if r < p: x.append(x[i] + 1) 
        else: x.append(x[i] - 1)

    trajetoria = np.array(x)

    return trajetoria
```

```python
x_vals = np.linspace(0, 1000, 1001)
plt.plot(x_vals, randomwalk(p=0.5, N=1000))
plt.xlabel("Passos")
plt.ylabel("Posição")
plt.title("Caminhada Aleatória")
plt.show()
```

```python
#2
    
def randomwalk_rapido(prob, N):

    r_numbers = rng.choice(a=[1,-1] , p=[prob, 1-prob], size=N, replace=True)
    x0 = [0]

    x = x0 + list(np.cumsum(r_numbers))

    return np.array(x)


plt.plot(x_vals, randomwalk_rapido(prob=0.5, N=1000))
plt.xlabel("Passos")
plt.ylabel("Posição")
plt.title("Caminhada Aleatória")
plt.show()
```

```python
#3

def randomwalk(prob, N, M):

    r_numbers = rng.choice(a=[1,-1] , p=[prob, 1-prob], size=(N,M), replace=True)
    x0 = np.zeros(M)

    trajetoria = np.vstack([x0, np.cumsum(r_numbers, axis=0)])

    return np.array(trajetoria)


```

```python
plt.plot(randomwalk(0.5, 1000, 500), color='red')
plt.plot(randomwalk(0.6, 1000, 500), color="blue")
plt.show()
```

```python
media1 = np.mean(a=randomwalk(0.5,1000,500), axis=1)
media2 = np.mean(a=randomwalk(0.6,1000,500), axis=1)

desvio1 = np.std(a=randomwalk(0.5,1000,500), axis=1)
desvio2 = np.std(a=randomwalk(0.6,1000,500), axis=1)

plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
plt.plot(media1, label="p=0.5", color="lightgreen")
plt.plot(media2, label="p=0.6", color="darkgreen")
plt.xlabel("Número de passos")
plt.ylabel("Média de todas as trajetórias")
plt.legend()

plt.subplot(1,2,2)
plt.plot(desvio1, label="p=0.5", color="purple")
plt.plot(desvio2, label="p=0.6", color="lightcoral")
plt.xlabel("Número de passos")
plt.ylabel("Desvio padrão de todas as trajetórias")
plt.legend()



plt.show()

```

```python
# 6) Histogramas em t=0, t=500 e t=1000

N = 1000
M = 500

traj1 = randomwalk(0.5, N, M)
traj2 = randomwalk(0.6, N, M)

plt.figure(figsize=(12,5))
plt.subplot(1,2,1)
plt.hist(traj1[0], bins=20, color="lightgreen", alpha=0.7)
plt.hist(traj1[500], bins=20, color="lightgreen", alpha=0.7)
plt.hist(traj2[1000], bins=20, color="lightgreen", alpha=0.7)

plt.title(f"Histograma p=0.5 em t=0,500,1000")
plt.xlabel("Posição")
plt.ylabel("Frequência")

plt.subplot(1,2,2)
plt.hist(traj1[10], bins=20, color="lightcoral", alpha=0.7)
plt.hist(traj1[500], bins=20, color="lightcoral", alpha=0.7)
plt.hist(traj2[1000], bins=20, color="lightcoral", alpha=0.7)

plt.title(f"Histograma p=0.5 em t=0,500,1000")
plt.xlabel("Posição")
plt.ylabel("Frequência")

    
plt.tight_layout()
plt.show()

plt.show()

```
